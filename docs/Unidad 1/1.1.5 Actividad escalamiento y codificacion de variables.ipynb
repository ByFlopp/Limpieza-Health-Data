##En este Notebook revisaremos lo visto en clase sobre escalamiento de variables

````python
import numpy as np
import pandas as pd

# Normalización y estandarización
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import RobustScaler
````

### Usaremos el data set de: California Housing

````python
# Puedes agregar la dirección que quieras, siempre y cuando haga referencia al dataset.
# Debes poner atención al separador. El parámetro "sep" indica como se separan los datos en el dataset.

url = "https://download.mlcc.google.com/mledu-datasets/california_housing_train.csv"
df_housing = pd.read_csv(url, sep=",")
df_housing
````

````python
# Algunas estadísticas básicas de solo las variables numéricas
df_housing.describe()
````

````python
# Revisamos el tipo de datos
df_housing.dtypes
````

#Estandarización

````python
# Se crea primero el objeto Standard Scaler
scaler = StandardScaler()

# Luego "entrenamos" con los datos para obtener los parámetros del escalamiento
scaler.fit(df_housing)

# Finalmente aplicamos el escalamiento
array_scaler = scaler.transform(df_housing)
array_scaler
````

### Se ve que entraga los datos más "desordenados". Esto pasa por que lo que entrega es un arreglo (array). Lo pasaremos a DataFrame

````python
df_standar_scaler = pd.DataFrame(array_scaler)
df_standar_scaler
````

**Ahora el problema es que las columnas perdieron sus nombres**

Podemos recuperar los nombres

````python
df_standar_scaler.columns = df_housing.columns
df_standar_scaler
````

Finalmente, corroboramos el resultado:

````python
df_standar_scaler.describe()
````

### ¿Cómo sabemos que lo hizo bien?

# Normalización

````python
# Se crea primero el objeto Standard Scaler
scaler = MinMaxScaler()

# Luego "entrenamos" con los datos para obtener los parámetros del escalamiento
scaler.fit(df_housing)

# Finalmente aplicamos el escalamiento
array_scaler = scaler.transform(df_housing)
array_scaler
````

````python
df_norm_scaler = pd.DataFrame(array_scaler)
df_norm_scaler
````

````python
df_norm_scaler.columns = df_housing.columns
df_norm_scaler
````

````python
df_norm_scaler.describe()
````

#Escalamiento Robusto

````python
robust_scaler = RobustScaler()
array_scaler = robust_scaler.fit_transform(df_housing)
df_robust_scaler = pd.DataFrame(array_scaler)
df_robust_scaler
````

````python
df_robust_scaler.columns = df_housing.columns
df_robust_scaler
````

````python
df_robust_scaler.describe()
````

#Ahora revisaremos lo visto en clase sobre encoding de variables categóricas

````python
import numpy as np
import pandas as pd

# Encodeo
from sklearn import preprocessing
````

````python
datos = {'alumno': ['Alejandro', 'Pedro', 'Sofia','Leonora' ,'Sergio', 'Maria'],
         'Primera prueba': [7, 8,7,7,8.5, 10],
         'Segunda prueba': [10,9,10,4,10,6.7],
         'Sexo':['Masculino','Masculino','Femenino','Femenino','Masculino','Femenino'],
         'Pais':['Chile', 'Chile', 'Peru', 'Brasil', 'Brasil','Chile'],
         'Año' :['Segundo', 'Tercero', 'Segundo', 'Primero', 'Tercero', 'Tercero']}
df = pd.DataFrame(datos)
df
````

#Ordinal Encoder

Aplicaremos Ordinal encoder a la variable "año". Ya que nos importa preservar el orden, usaremos la función .replace(). Es muy sencillo hacerlo de esta manera.

````python
# Usaremos la función replace, ya que es más facil.
df['Año'] = df['Año'].replace({'Primero': 1, 'Segundo': 2, 'Tercero': 3})
df
````

#One-Hot Encoder

Aplicaremos One-hot encoder a la variable categórica Sexo.

````python
# Aplicar One-Hot Encoding usando get_dummies() de pandas
#data = df['Sexo']
ohe = pd.get_dummies(df, columns=['Sexo'])
ohe
````

Como se puede observar, se crean 2 variables, una por cada valor de la variable "sexo". Pero estan deben tener valores numéricos. Por lo tanto, primero les daremos un valor numérico y luego aplicaremos un encoding del tipo One-hot

````python
# Usaremos la función replace, ya que es más facil.
df['Sexo'] = df['Sexo'].replace({'Masculino': '0', 'Femenino': '1'})
df
````

````python
ohe = pd.get_dummies(df, columns=['Sexo'], dtype=int)
ohe
````

Solo falta cambiar los nombres de las columnas. Esto es opcional.

````python
ohe.columns = ['alumno','Primera prueba', 'Segunda prueba',	'Pais',	'Año', 'Masculino', 'Femenino']
ohe
````

````python
# Con esto revisamos el tipo de dato de las variables y corroboramos que sea numérico.
ohe.dtypes
````
